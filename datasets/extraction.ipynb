### This noteboook is to do Batch inference with the NER model trained on the big clean dataset that has 85k instances.

#### Necessary Libraries

```python
import spacy
import pandas as pd
import os
```

#### Inference Code

```python
model_path = os.path.join(os.getcwd(), "transformer-models/model-best")
nlp = spacy.load(model_path)

data_path = os.path.join(os.getcwd(), "NER_ready_data/combined.csv")
df = pd.read_csv(data_path, encoding="utf-8")

texts = df["text"].dropna().tolist()
print(f"There are {len(texts)} resumes in the dataset")

doc_counter = 0
extracted_data = []

for doc in nlp.pipe(texts, batch_size=50):
    doc_counter += 1
    if doc_counter % 1000 == 0:
        print(f"processed {doc_counter} / {len(texts)} entries...")
    
    for ent in doc.ents:
        extracted_data.append({
            "doc_id": doc_counter,
            "text": ent.text,
            "label": ent.label_       
         })

print("Docs are labelled without issue")
df_labelled = pd.DataFrame(extracted_data)
df_labelled.to_csv("raw_entities.csv", index=False, encoding="utf-8")
print(f"labels are saved and ready for the distillation")
```
